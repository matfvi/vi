```python
import copy
from itertools import product
import math
from collections import defaultdict

def load(filename):
    with open(filename, 'r') as f:
        return [line.strip() for line in f.readlines()]

def print_teren(t):
    for red in t:
        print(red)

def find(teren, name):
    n = len(teren)
    m = len(teren[0])
    for i in range(n):
        for j in range(m):
            if teren[i][j] == name:
                return (i, j)


def astar(teren, start, target, h):
    iterations = 0
    path = []
    n = len(teren)
    m = len(teren[0])
    
    # POCETAK STUDENTSKOG KODA
    open_set, closed_set = set([start]), set()

    cheapest_paths = defaultdict(lambda: float('inf'))
    cheapest_paths[start] = 0

    parent = {start: None}

    def get_best():
        result, result_value = None, float('inf')
        for node in open_set:
            new_value = cheapest_paths[node] + h(node, target)
            if not result or new_value < result_value:
                result, result_value = node, new_value
        return result

    path_found = False
    while len(open_set) > 0:
        current = get_best()
        iterations += 1
        if current == target:
            path_found = True
            break

        i, j = current
        for neighbour in ((i - 1, j), (i + 1, j), (i, j - 1), (i, j + 1)):
            ni, nj = neighbour
            if not (0 <= ni < n and 0 <= nj < m):
                continue
            if teren[ni][nj] == '#':
                continue

            new_path_to_neighbour_weight = cheapest_paths[current] + 1
            if neighbour not in open_set and neighbour not in closed_set:
                open_set.add(neighbour)
                parent[neighbour] = current
                cheapest_paths[neighbour] = new_path_to_neighbour_weight
            elif new_path_to_neighbour_weight < cheapest_paths[neighbour]:
                parent[neighbour] = current
                cheapest_paths[neighbour] = new_path_to_neighbour_weight
                if neighbour in closed_set:
                    closed_set.remove(neighbour)
                    open_set.add(neighbour)

        open_set.remove(current)
        closed_set.add(current)

    if path_found:
        while target is not None:
            path.append(target)
            target = parent[target]
        path.reverse()
    # KRAJ STUDENTSKOG KODA
    
    return iterations, path


def euclidean(a, b):
    return math.sqrt((a[0] - b[0]) ** 2 + (a[1] - b[1]) ** 2)


def manhattan(a, b):
    return abs(a[0] - b[0]) + abs(a[1] - b[1])


def chebyshev(a, b):
    return max(abs(a[0] - b[0]), abs(a[1] - b[1]))

if __name__ == '__main__':
    #POCETAK STUDENTSKOG KODA
    teren = load('teren.txt')
    print_teren(teren)
    A_coords = find(teren, 'A')
    B_coords = find(teren, 'B')
    print(A_coords, B_coords)

    results = []
    for name, heuristic in [
        ('Euklidska', euclidean),
        ('Menhetn', manhattan),
        ('Cebisevljeva', chebyshev),
    ]:
        iterations, path = astar(teren, A_coords, B_coords, heuristic)
        results.append((iterations, name, path))
        print(f'{name}: iteracije={iterations}, duzina_puta={len(path)}')

    best_iterations, best_name, best_path = min(results, key=lambda item: item[0])
    print('Najbolja heuristika:', best_name)
    print('Broj iteracija:', best_iterations)
    print('Put:', best_path)
    # KRAJ STUDENTSKOG KODA
```
