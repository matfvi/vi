```python
from z3 import *
# DeclareSort, BoolSort, Function, ForAll, Exists, Solver, Const, Consts, And, Or, Implies
```

```python
Person = DeclareSort('Person')

Invited = Function('Invited', Person, BoolSort())
InPhonebook = Function('InPhonebook', Person, Person, BoolSort())
friend = Function('friend', Person, BoolSort())

nikola, stefan, nemanja, andrija, ivan, dragan = Consts('Nikola Stefan Nemanja Andrija Ivan Dragan', Person)
x, y, z = Consts('x y z', Person)

solver = Solver()
axioms = [
    Distinct(nikola, stefan, nemanja, andrija, ivan, dragan),

    friend(stefan),
    friend(nemanja),
    friend(andrija),
    friend(ivan),
    friend(dragan),

    InPhonebook(nikola, ivan),
    InPhonebook(stefan, ivan),
    InPhonebook(nemanja, dragan),
    InPhonebook(nemanja, ivan),
    InPhonebook(nemanja, nikola),
    InPhonebook(dragan, nemanja),
    InPhonebook(dragan, andrija),
    InPhonebook(ivan, nemanja),
    InPhonebook(ivan, stefan),
    InPhonebook(andrija, nikola),

    Invited(ivan),
    ForAll([x, y], Implies(And(Invited(x), InPhonebook(x, y)), Invited(y))),
]

solver.add(axioms)

if solver.check() == sat:
    conjecture = And(Invited(stefan), Invited(nemanja), Invited(andrija), Invited(ivan), Invited(dragan))
    solver.add(Not(conjecture))
    result = solver.check()
    print(result)
    if result == unsat:
        print('Conjecture proved!')
    else:
        print(solver.model())
else:
    print('contradicting axioms')
```
